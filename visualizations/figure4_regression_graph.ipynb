# Figure 4: Linear Regression Forward Feature Selection

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Times New Roman'
```

```python
# Load feature selection scores (run feature_selection_and_importance.ipynb first)

fval_cv = np.load('./data/FValue_cv_scores.npy')
fval_train = np.load('./data/FValue_train_scores.npy')

w1_cv = np.load('./data/W1_cv_scores.npy')
w1_train = np.load('./data/W1_train_scores.npy')
```

```python
fig, axs = plt.subplots(1, 2, figsize=(14, 5))

# phi_tau
axs[0].plot(np.arange(1, len(fval_train)+1), fval_train, label='Training', color='tab:blue')
axs[0].plot(np.arange(1, len(fval_cv)+1), fval_cv, label='Cross Validation', color='tab:green')
axs[0].axvline(x=10, color='red', linestyle='--', label='Model Selection Elbow Point')
axs[0].set_xlabel('Number of Regression Model Features', fontsize=14)
axs[0].set_ylabel('NMSE', fontsize=14)
axs[0].set_title(r'$\phi_{\tau}$', fontsize=16)
axs[0].set_ylim(0, 1.2)

# Delta_tau
axs[1].plot(np.arange(1, len(w1_train)+1), w1_train, label='Training', color='tab:blue')
axs[1].plot(np.arange(1, len(w1_cv)+1), w1_cv, label='Cross Validation', color='tab:green')
axs[1].axvline(x=10, color='red', linestyle='--', label='Model Selection Elbow Point')
axs[1].set_xlabel('Number of Regression Model Features', fontsize=14)
axs[1].set_title(r'$\Delta_{\tau}$', fontsize=16)
axs[1].set_ylim(0, 1.2)
axs[1].legend(fontsize=11)

fig.suptitle('Linear regression forward feature selection', fontsize=18, fontweight='bold')
plt.tight_layout()
plt.savefig('Figure4.png', dpi=300, bbox_inches='tight')
plt.show()
```
